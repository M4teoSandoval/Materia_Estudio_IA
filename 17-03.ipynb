```python
import numpy as np

# Declarar la matriz de 1s de 3x3, convertirla a float y dividir entre 9
kernel_suavizar = np.ones((3, 3), dtype=np.float32) / 9

# Mostrar cómo quedó la matriz
print(kernel_suavizar)
```

```python
import cv2
# import matplotlib.pyplot as plt # Descomenta esto si quieres visualizar en Jupyter

# Asumiendo que tienes una imagen cargada llamada 'imagen'
imagen = cv2.imread('logo.jpg')

# Aplicar el filtro 2D
# ddepth=-1 indica que la imagen de salida tendrá la misma profundidad (tipo de dato) que la imagen original
imagen_suavizada = cv2.filter2D(src=imagen, ddepth=-1, kernel=kernel_suavizar)

# Código opcional para mostrar la imagen original y la suavizada en Jupyter Notebook

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
axes[0].set_title('Imagen Original')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(imagen_suavizada, cv2.COLOR_BGR2RGB))
axes[1].set_title('Imagen Suavizada (Kernel 3x3)')
axes[1].axis('off')

plt.show()

```

```python
import numpy as np

# 1. Kernel Sobel Vertical (Detector de bordes verticales, derivada en X)
# Formado por la aproximación de la derivada horizontal multiplicada por un suavizado vertical
kernel_sobel_vertical = np.array([
    [-1., -2.,  0.,  2.,  1.],
    [-4., -8.,  0.,  8.,  4.],
    [-6.,-12.,  0., 12.,  6.],
    [-4., -8.,  0.,  8.,  4.],
    [-1., -2.,  0.,  2.,  1.]
], dtype=np.float32)

# 2. Kernel Sobel Horizontal (Detector de bordes horizontales, derivada en Y)
# Formado por la aproximación de la derivada vertical multiplicada por un suavizado horizontal
kernel_sobel_horizontal = np.array([
    [-1., -4., -6., -4., -1.],
    [-2., -8.,-12., -8., -2.],
    [ 0.,  0.,  0.,  0.,  0.],
    [ 2.,  8., 12.,  8.,  2.],
    [ 1.,  4.,  6.,  4.,  1.]
], dtype=np.float32)

# 3. Kernel Gaussiano 5x5 (Filtro de desenfoque suave)
# Aproximación entera estándar, la suma de todos los valores da 256, por lo que se divide entre 256
kernel_gaussiano = np.array([
    [1.,  4.,  6.,  4.,  1.],
    [4., 16., 24., 16.,  4.],
    [6., 24., 36., 24.,  6.],
    [4., 16., 24., 16.,  4.],
    [1.,  4.,  6.,  4.,  1.]
], dtype=np.float32) / 256.0

# 4. Kernel de Repujado 5x5 (Emboss)
# Efecto de relieve creando una sombra en diagonal (valores negativos arriba-izquierda, positivos abajo-derecha)
kernel_repujado = np.array([
    [-2., -2., -2., -1.,  0.],
    [-2., -2., -1.,  0.,  1.],
    [-2., -1.,  1.,  1.,  2.],
    [-1.,  0.,  1.,  2.,  2.],
    [ 0.,  1.,  2.,  2.,  2.]
], dtype=np.float32)

# ====== MOSTRAR EN PANTALLA ======
print("--- Kernel Sobel Vertical 5x5 ---")
print(kernel_sobel_vertical)
print("\n--- Kernel Sobel Horizontal 5x5 ---")
print(kernel_sobel_horizontal)
print("\n--- Kernel Gaussiano 5x5 ---")
print(kernel_gaussiano)
print("\n--- Kernel Repujado (Emboss) 5x5 ---")
print(kernel_repujado)
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. Aplicar los filtros a la imagen 'img2' usando el método cv2.filter2D
# ddepth=-1 mantendrá la misma profundidad de color de la imagen original
img_sobel_v = cv2.filter2D(src=img2, ddepth=-1, kernel=kernel_sobel_vertical)
img_sobel_h = cv2.filter2D(src=img2, ddepth=-1, kernel=kernel_sobel_horizontal)
img_gauss = cv2.filter2D(src=img2, ddepth=-1, kernel=kernel_gaussiano)
img_repujado = cv2.filter2D(src=img2, ddepth=-1, kernel=kernel_repujado)

# 2. Configurar la visualización de las imágenes
# OpenCV por defecto lee en formato BGR. Esta función auxiliar las cambia a RGB
# para que los colores se vean correctamente con matplotlib.
def to_rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Creamos una cuadrícula de 2 filas y 3 columnas para mostrar las 5 imágenes
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# [0, 0] Imagen original
axes[0, 0].imshow(to_rgb(img2))
axes[0, 0].set_title('1. Original (img2)', fontsize=14)
axes[0, 0].axis('off')

# [0, 1] Imagen Sobel Vertical
axes[0, 1].imshow(to_rgb(img_sobel_v))
axes[0, 1].set_title('2. Sobel Vertical (Bordes L-R)', fontsize=14)
axes[0, 1].axis('off')

# [0, 2] Imagen Sobel Horizontal
axes[0, 2].imshow(to_rgb(img_sobel_h))
axes[0, 2].set_title('3. Sobel Horizontal (Bordes U-D)', fontsize=14)
axes[0, 2].axis('off')

# [1, 0] Imagen Gaussiana
axes[1, 0].imshow(to_rgb(img_gauss))
axes[1, 0].set_title('4. Suavizado Gaussiano', fontsize=14)
axes[1, 0].axis('off')

# [1, 1] Imagen Repujado
axes[1, 1].imshow(to_rgb(img_repujado))
axes[1, 1].set_title('5. Repujado (Emboss)', fontsize=14)
axes[1, 1].axis('off')

# [1, 2] Ocultar el último recuadro de la cuadrícula, ya que solo tenemos 5 imágenes
axes[1, 2].axis('off')

# Ajustar los espacios y mostrar el resultado en pantalla
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. APLICAR MAX POOLING A LA IMAGEN 'img2' ---

# Obtener las nuevas dimensiones de img2
H, W, C = img2.shape

# Aseguramos que el alto y ancho sean pares recortando 1 píxel si es impar
H_par = H - (H % 2)
W_par = W - (W % 2)
img_recortada = img2[:H_par, :W_par, :]

# Aplicamos el Max Pooling a img2 y lo guardamos en una nueva variable (ej: img3)
img3 = img_recortada.reshape(H_par//2, 2, W_par//2, 2, C).max(axis=(1, 3)).astype(np.uint8)

# --- 2. MOSTRAR EN PANTALLA EL RESULTADO ---

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Función para colorear bien la imagen si fue leída con cv2 original
def to_rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Mostrar img2 (la víctima que vas a reducir de nuevo)
axes[0].imshow(to_rgb(img2))
axes[0].set_title(f'Primer Max Pooling (img2)\nTamaño: {img2.shape}', fontsize=12)
axes[0].axis('off')

# Mostrar img3 (resultado del segundo Max Pooling)
axes[1].imshow(to_rgb(img3))
axes[1].set_title(f'Segundo Max Pooling (img3)\nTamaño: {img3.shape}', fontsize=12)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. APLICAR MAX POOLING 3x3 A 'img_sobel_v' ---

# Obtener dimensiones
H, W, C = img_sobel_v.shape

# Para un pooling de 3x3, el alto y el ancho deben ser múltiplos perfectos de 3.
# Calculamos exactamente cuánto sobra con % 3 y lo recortamos:
H_multiplo3 = H - (H % 3)
W_multiplo3 = W - (W % 3)
img_recortada = img_sobel_v[:H_multiplo3, :W_multiplo3, :]

# Reagrupamos en bloques de (3x3) y extraemos el valor máximo.
# Tu nueva imagen medirá una tercera parte de la original en ancho y largo.
img_sobel_pool3x3 = img_recortada.reshape(H_multiplo3//3, 3, W_multiplo3//3, 3, C).max(axis=(1, 3)).astype(np.uint8)

# --- 2. MOSTRAR EN PANTALLA EL RESULTADO ---

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

def to_rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Mostrar la imagen original con el filtro de Sobel
axes[0].imshow(to_rgb(img_sobel_v))
axes[0].set_title(f'Sobel Vertical\nTamaño: {img_sobel_v.shape}', fontsize=12)
axes[0].axis('off')

# Mostrar el resultado del Max Pooling 3x3
axes[1].imshow(to_rgb(img_sobel_pool3x3))
axes[1].set_title(f'Sobel Vert. + Max Pooling 3x3\nTamaño: {img_sobel_pool3x3.shape}', fontsize=12)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 1. APLICAR EL FILTRO SOBEL 5x5 A 'img_sobel_pool3x3' ---
# Usaremos el kernel_sobel_vertical declarado anteriormente
img_sobel_pool3x3_sobel = cv2.filter2D(src=img_sobel_pool3x3, ddepth=-1, kernel=kernel_sobel_vertical)


# --- 2. APLICAR MAX POOLING 2x2 ---

# Obtener las nuevas dimensiones de la imagen con el filtro
H, W, C = img_sobel_pool3x3_sobel.shape

# Aseguramos que el alto y ancho sean pares recortando 1 píxel si sobra
H_par = H - (H % 2)
W_par = W - (W % 2)
img_recortada = img_sobel_pool3x3_sobel[:H_par, :W_par, :]

# Aplicamos el Max Pooling 2x2 y lo guardamos en una nueva variable (ej: img4)
img4 = img_recortada.reshape(H_par//2, 2, W_par//2, 2, C).max(axis=(1, 3)).astype(np.uint8)


# --- 3. MOSTRAR EN PANTALLA LOS RESULTADOS ---

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

def to_rgb(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Mostrar img_sobel_pool3x3 original (la que entró a la capa actual)
axes[0].imshow(to_rgb(img_sobel_pool3x3))
axes[0].set_title(f'Entrada (img_sobel_pool3x3)\nTamaño: {img_sobel_pool3x3.shape}', fontsize=12)
axes[0].axis('off')

# Mostrar img4 (el resultado tras Filtro Sobel + Pooling)
axes[1].imshow(to_rgb(img4))
axes[1].set_title(f'Salida (Filtro Sobel 5x5 + Pooling 2x2)\nTamaño: {img4.shape}', fontsize=12)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
